```python
#2️⃣ Count vowels and consonants
Input = input("Enter a Sentense:")
vowels = 0
consonants = 0
for word in Input:
    if word in "aeiouAEIOU":
        vowels += 1
    else:
        consonants += 1
print("Vowels:",vowels)
print("Consonants:",consonants)
```

```python
Dictionary

5️⃣ Count frequency of characters in string
```

```python
#5️⃣ Count frequency of characters in string
Input = input("Enter a Sentense:")
d = {}
Words = Input.split()
for char in Words:
    if char in d:
        d[char] += 1
    else:
        d[char] = 1
print("Character Frequency:",d)
```

```python
Mini Data Analyst Task 📊

10️⃣ Given:

sales = [1200,1500,1100,1800,2000,1700]

Find:

total sales
average sales
highest sale
lowest sale
```

```python
sales = [1200,1500,1100,1800,2000,1700]
total_sales = 0
for num in sales:
    total_sales += num
average_sales = total_sales/len(sales)
highest_sale = max(sales)
lowest_sale = min(sales)
print("Highest sale:",highest_sale)
print("Lowest sale:",lowest_sale)
print("Average sales:",average_sales)
print("total sales:",total_sales)
```

```python
4️⃣ Write a program to calculate the factorial of a number.

5️⃣ Write a program to check whether a number is prime or not.

8️⃣ Write a program to generate the Fibonacci series up to n terms.
```

```python
#4️⃣ Write a program to calculate the factorial of a number.
User = int(input("Enter a number:"))
Factorial = 1
for i in range(1,User+1):
    Factorial *=  i
print(f"The Factorial of {User} is {Factorial}")
```

```python
#5️⃣ Write a program to check whether a number is prime or not.
User = int(input("Enter a number:"))
if User <= 1:
    print(f"Sorry {User} is not Prime number")
else:
    is_prime = True
    for i in range(2,int(User**0.5)+1):
        if User % i == 0:
            is_prime = False
            break

    if is_prime:
        print(f"{User} is a prime number.")
    else:
        print(f"{User} is not a prime number.")
```

```python
#8️⃣ Write a program to generate the Fibonacci series up to n terms.
nth = int(input("Enter a number:"))
a = 0
b = 1

print(f"Fibonacci series up to {nth} terms:",end=" ")

for i in range(nth):
    print(a, end=" ")
    a,b = b, a + b
```
